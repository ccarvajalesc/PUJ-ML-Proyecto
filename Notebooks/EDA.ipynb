```python
#!/usr/bin/env python

# make sure to install these packages before running:
# pip install pandas
# pip install sodapy

import pandas as pd
from sodapy import Socrata

# Unauthenticated client only works with public data sets. Note 'None'
# in place of application token, and no username or password:
client = Socrata(
    "www.datos.gov.co",
    None,
    timeout=120
)

# Example authenticated client (needed for non-public datasets):
# client = Socrata(www.datos.gov.co,
#                  MyAppToken,
#                  username="user@example.com",
#                  password="AFakePassword")

# First 2000 results, returned as JSON from API / converted to Python list of
# dictionaries by sodapy.
results = client.get("kgxf-xxbe", limit=2000)

# Convert to pandas DataFrame
results_df = pd.DataFrame.from_records(results)
```

```python
periodos = client.get(
    "kgxf-xxbe",
    select="distinct periodo",
    order="periodo DESC",
    limit=100
)

pd.DataFrame(periodos)
```

```python
results = client.get(
    "kgxf-xxbe",
    where="periodo = '20224'",
    limit=10000
)
```

```python
# ─── Cantidad de registros de un período específico ──────────────────────────

periodo_consulta = '20224'

resultado = client.get(
    "kgxf-xxbe",
    select="count(*) as total_registros",
    where=f"periodo = '{periodo_consulta}'"
)

# Extraer cantidad
total = int(resultado[0]['total_registros'])

print(f"📊 Total de registros para el período {periodo_consulta}: {total:,}")
```
